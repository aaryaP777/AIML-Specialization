```python
import pandas as pd
import numpy as np
import re
import nltk
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity, euclidean_distances
from collections import Counter
```

```python
# Download required NLTK resources
nltk.download('punkt')
```

```python
# Sample Dataset (Title, Author, Words)
documents = [
    {"T": "18 Editions of the Dewey Decimal Classifications",
     "A": "Comaromi, J.P.",
     "W": "History of the DEWEY Decimal Classification and its impact."},

    {"T": "Use Made of Technical Libraries",
     "A": "Slater, M.",
     "W": "Analysis of library use in technical environments."},

    {"T": "Two Kinds of Power",
     "A": "Wilson, P.",
     "W": "Bibliographic control as a form of knowledge power."}
]

# Convert to DataFrame
df = pd.DataFrame(documents)
```

```python
# Convert to DataFrame
df = pd.DataFrame(documents)

# Preprocessing: Combine title and words for representation
df['text'] = df['T'] + " " + df['W']

# Display Dataset
df.head()
```

```python
# TF-IDF Representation
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(df['text'])

# Convert to DataFrame for readability
df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=vectorizer.get_feature_names_out())

# Display TF-IDF Matrix
df_tfidf
```
