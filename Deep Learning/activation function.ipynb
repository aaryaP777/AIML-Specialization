```python
import numpy as np 
import matplotlib.pyplot as plt
```

Sigmoid func.

```python
def sigmoid(x):
    '''It returns 1/(1+exp(-x)). where the values lies between zero and one '''
    
    return 1/(1+np.exp(-x))
```

```python
x = np.linspace(-10, 10)
plt.plot(x, sigmoid(x))
plt.axis('tight')
plt.title('Activation Function :Sigmoid')
plt.show()
```

TanH func.

```python
def tanh(x):
    return np.tanh(x)

#It returns the value (1-exp(-2x))/(1+exp(-2x)) and the value returned will be lies in between -1 to 1.
```

```python
x = np.linspace(-10, 10)
plt.plot(x, tanh(x))
plt.axis('tight')
plt.title('Activation Function :Tanh')
plt.show()
```

RELU

```python
def relu(x):
    
    x1 = []
    for i in x:
        if i<0:
            x1.append(0)
        else:
            x1.append(i)
            
    return x1

# It returns zero if value is < 0, otherwise returns the given input.
```

```python
x = np.linspace(-10, 10)
plt.plot(x, relu(x))
plt.axis('tight')
plt.title('Activation Function :RELU')
plt.show()
```

Leaky Relu

```python
def leaky_relu(x, a):
    x1=[]
    for i in x:
        if i<0:
            x1.append(a*i)
        else:
            x1.append(i)
            
    return x1

# It returns zero if the input is less than zero otherwise it returns the given input.
```

```python
x = np.linspace(-10, 10)
plt.plot(x, leaky_relu(x,0.1))
plt.axis('tight')
plt.title('Activation Function :Leaky RELU')
plt.show()
```

Softmax

```python
def softmax(x):
    return np.exp(x)/np.sum(np.exp(x), axis=0)

# Compute softmax values for each sets of scores in x.
```

```python
x = np.linspace(-10, 10)
plt.plot(x, softmax(x))
plt.axis('tight')
plt.title('Activation Function :Softmax')
plt.show()
```

Derivatives of activation functions

```python
# sigmoidal derivative

def dsig(x):
    
    return sigmoid(x) * (1-sigmoid(x))
```

```python
# Hyperbolic derivative

def hyp(x):
    return (np.exp(x) - np.exp(-x))/(np.exp(x) + np.exp(-x))

def dhyp(x):
    return 1 - hyp(x) * hyp(x)
```

```python
def drelu(x):
    temp = [1 if value>0 else 0 for value in x]
    return np.array(temp, dtype = float)
```

```python
def dLrelu(x,a):
    temp = [1 if value>0 else a for value in x]
    return np.array(temp, dtype = float)
```

```python
x=np.linspace(-10, 10)
labels=['X','Sig','Tanh','Relu','LRelu']
plt.plot(x,dsig(x),label=labels[1])
plt.plot(x,dhyp(x),label=labels[2])
plt.plot(x,drelu(x),label=labels[3])
plt.plot(x,dLrelu(x,0.1),label=labels[4])

plt.title('Derivatives of Activation Function')
plt.legend()
plt.show()
```
