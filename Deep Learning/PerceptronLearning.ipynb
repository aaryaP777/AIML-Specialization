```python
import numpy as np
```

```python
class Perceptron:
    def __init__(self, W, learning_rate=0.01, n_iters=1000):
        self.lr = learning_rate
        self.n_iters = n_iters
        self.activation_func = self._unit_step_func
        self.weights = W #None
        self.bias = None

    def fit(self, X, y):

        # init parameters
        self.bias = 0

        y_ = np.array([1 if i > 0 else 0 for i in y])

        linear_output = np.dot(np.transpose(self.weights) , X) + self.bias
        print(linear_output)
        y_predicted = self.activation_func(linear_output)

       # Perceptron update rule
        update = self.lr * (y - y_predicted)
        print(self.weights.shape)
        self.weights = self.weights + update * X
        self.bias += update

        print(self.weights)

    def _unit_step_func(self, x):
        return np.where(x >= 0, 1, 0)
```

```python
w=[1,-1,0,0.5]
wa=np.array(w)
wi=wa.reshape(4,1)
print(wi)
print(wi.shape)
```

```python
#Testing
p = Perceptron(wi, learning_rate=0.01, n_iters=1)
```
