```python
def get_fuzzy_set(name):
    fuzzy_set = {}
    n = int(input(f"Enter the number of elements in the fuzzy set {name}: "))
    for _ in range(n):
        key = input("Enter the element: ").strip()
        while True:
            try:
                value = float(input(f"Enter membership value for {key} (between 0 and 1): "))
                if 0 <= value <= 1:
                    break
                else:
                    print("Invalid input! Membership value must be between 0 and 1.")
            except ValueError:
                print("Invalid input! Please enter a numeric value between 0 and 1.")
        fuzzy_set[key] = value
    return fuzzy_set

def is_subset(A, B):
    for key in A:
        if key not in B or A[key] > B[key]:
            return False
    return True

A = get_fuzzy_set("A")
B = get_fuzzy_set("B")

print('\nA = ', A)
print('B = ', B)

Y = {} #UNION
Z = {} #INTERSECTION
XA = {} #COMPLEMENT A
XB = {} #COMPLEMENT B

#UNION
for Akey, Bkey in zip(A,B):
    Avalue = A[Akey]
    Bvalue = B[Bkey]
    
    if Avalue > Bvalue:
        Y[Akey] = Avalue
    else:
        Y[Bkey] = Bvalue
print('\nUnion is: ', Y)

#INTERSECTION
for Akey, Bkey in zip(A,B):
    Avalue = A[Akey]
    Bvalue = B[Bkey]
    
    if Avalue < Bvalue:
        Z[Akey] = Avalue
    else:
        Z[Bkey] = Bvalue
print('\nIntersection is: ', Z) 

#COMPLEMENT
for Akey in A:
    XA[Akey] = round(1 - A[Akey], 2)
print('\nComplement of A is: ', XA)

for Bkey in B:
    XB[Bkey] = round(1 - B[Bkey], 2)
print('\nComplement of B is: ', XB)

# COMPLEMENT OF UNION
comp_union = {key: round(1 - Y[key]) for key in Y}
    
# COMPLEMENT OF INTERSECTION
comp_inter = {key: round(1 - Z[key]) for key in Z}

# INTERSECTION OF COMPLEMENT
inter_comps = {key: min(XA.get(key, 1), XB.get(key, 1)) for key in set(A.keys()).union(B.keys())}

# UNION OF COMPLEMENT
union_comps = {key: max(XA.get(key, 0), XB.get(key, 0)) for key in set(A.keys()).union(B.keys())}

#SUBSET
if is_subset(A, B):
    print("\nA is a fuzzy subset of B")
else:
    print("\nA is NOT a fuzzy subset of B")
       
#DE-MORGAN'S LAW
print("\nDe Morgan's Laws Verification:")
print("LHS (Complement of Union):", comp_union)
print("RHS (Intersection of Complements):", inter_comps)
print("LHS (Complement of Intersection):", comp_inter)
print("RHS (Union of Complements):", union_comps)

if comp_union == inter_comps:
     print("\nDe Morgan's Law 1 is Verified: Complement of Union equals Intersection of Complements.")
else:
    print("\nDe Morgan's Law 1 is NOT Verified: There is a discrepancy.")
    
if comp_inter == union_comps:
    print("De Morgan's Law 2 is Verified: Complement of Intersection equals Union of Complements.")
else:
    print("De Morgan's Law 2 is NOT Verified: There is a discrepancy.")
```
