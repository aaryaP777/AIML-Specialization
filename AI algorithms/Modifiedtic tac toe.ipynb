```python
import random
import numpy as np

def print_board(board):
    for row in board:
        print(" ".join(row))
    print()

def is_board_full(board):
    for row in board:
        if '_' in row:
            return False
    return True

def check_winner(board):
    # Check rows
    if board[0][0] == board[0][1] == board[0][2] == board[0][3] != '_':
        return board[0][0]
    for i in range(1, 4):
        if board[i][0] == board[i][1] == board[i][2] != '_':
            return board[i][0]
    
    # Check columns
    for j in range(3):
        if board[1][j] == board[2][j] == board[3][j] != '_':
            return board[1][j]
    for j in range(4):
        if board[0][j] == board[1][min(j,2)] == board[2][min(j,2)] != '_':
            return board[0][j]
    
    # Check diagonals
    if board[0][0] == board[1][1] == board[2][2] != '_':
        return board[0][0]
    if board[0][3] == board[1][2] == board[2][1] != '_':
        return board[0][3]
    return None

def best_move(board, ai_symbol, human_symbol):
    best_score = -np.inf
    move = None
    for i in range(4):
        for j in range(len(board[i])):
            if board[i][j] == '_':
                board[i][j] = ai_symbol
                score = minimax(board, 0, False, ai_symbol, human_symbol, -np.inf, np.inf)
                board[i][j] = '_'
                if score > best_score:
                    best_score = score
                    move = (i, j)
    return move

def minimax(board, depth, is_maximizing, ai_symbol, human_symbol, alpha, beta):
    winner = check_winner(board)
    if winner == ai_symbol:
        return 10 - depth  # Prioritize faster wins
    elif winner == human_symbol:
        return depth - 10  # Delay opponent win
    elif is_board_full(board):
        return 0
    
    if depth >= 5:  # Limit search depth for faster decisions
        return 0
    
    if is_maximizing:
        best_score = -np.inf
        for i in range(4):
            for j in range(len(board[i])):
                if board[i][j] == '_':
                    board[i][j] = ai_symbol
                    score = minimax(board, depth + 1, False, ai_symbol, human_symbol, alpha, beta)
                    board[i][j] = '_'
                    best_score = max(score, best_score)
                    alpha = max(alpha, best_score)
                    if beta <= alpha:
                        break  # Prune branches
        return best_score
    else:
        best_score = np.inf
        for i in range(4):
            for j in range(len(board[i])):
                if board[i][j] == '_':
                    board[i][j] = human_symbol
                    score = minimax(board, depth + 1, True, ai_symbol, human_symbol, alpha, beta)
                    board[i][j] = '_'
                    best_score = min(score, best_score)
                    beta = min(beta, best_score)
                    if beta <= alpha:
                        break  # Prune branches
        return best_score

def play_game():
    board = [['' for _ in range(4)]] + [['' for _ in range(3)] for _ in range(3)]
    print("Tossing a coin...")
    toss = random.choice(['Head', 'Tail'])
    print(f"{toss} wins!")
    
    if toss == 'Head':
        ai_symbol = random.choice(['X', 'O'])
        human_symbol = 'O' if ai_symbol == 'X' else 'X'
        turn = 'Computer'
    else:
        human_symbol = input("Choose X or O: ").upper()
        ai_symbol = 'O' if human_symbol == 'X' else 'X'
        turn = 'Human'
    
    print_board(board)
    while True:
        if turn == 'Human':
            try:
                row, col = map(int, input("Enter row and column (1-4): ").split())
                row, col = row - 1, col - 1
                if 0 <= row < 4 and 0 <= col < len(board[row]) and board[row][col] == '_':
                    board[row][col] = human_symbol
                    turn = 'Computer'
                else:
                    print("Invalid move! Try again.")
                    continue
            except:
                print("Invalid input! Enter valid row and column.")
                continue
        else:
            print("Computer is making a move...")
            move = best_move(board, ai_symbol, human_symbol)
            if move:
                board[move[0]][move[1]] = ai_symbol
            turn = 'Human'
        
        print_board(board)
        winner = check_winner(board)
        if winner:
            print(f"{winner} wins!")
            break
        if is_board_full(board):
            print("It's a draw!")
            break

if _name_ == "_main_":
    play_game()
```
