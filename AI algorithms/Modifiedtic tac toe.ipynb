```python
import random 

board = ["" for x in range(10)]

def print_board():
    row1 = "|  {}  |  {}  |  {}  |  {}  |".format(board[0], board[1], board[2], board[3])
    row2 = "|  {}  |  {}  |  {}  |".format(board[4], board[5], board[6])
    row3 = "|  {}  |  {}  |  {}  |".format(board[7], board[8], board[9])
    
    print()
    print(row1)
    print(row2)
    print(row3)
    print()

def toss():
    result = random.choice(["heads", "tails"])
    print(f"Toss result: {result}")
    
    if result == "heads":
        print("AI plays!")
        return 1
    
    else:
        print("Human plays!")
        return 2
       
toss()
        
```
